```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\Farman Ali\Documents\semester 3\ICA\ride_hailing_dataset.csv")
df
```

## Average Fare by Service Type

```python
avg_fare = df.groupby("service_type")["fare"].mean()

avg_fare
```

## Which city has most Trips?

```python
df.groupby("city")["trip_id"].count().sort_values(ascending=False).head(1)
```

## Average Distance per Trip

```python
print(f"Avg Distance per trip: {df["distance_km"].mean():.2f} KM")
```

## Average Trip Rating by Gender

```python
ratings = df.groupby("gender")["trip_rating"].mean()

print(f"Average Trip Rating by Male: {ratings["Male"]:.2f}")
print(f"Average Trip Rating by Female: {ratings["Female"]:.2f}")

if (ratings["Female"] > ratings["Male"]):
    print("Female Trip Rating is better!")
elif ratings["Female"] < ratings["Male"]:
    print("Male Trip Rating is better!")
else:
    print("Male and Female trip rating is same")
```

## Service Type and Average Fare

```python
avg_fare = df.groupby("service_type")["fare"].mean()



fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(avg_fare.index.astype(str), avg_fare.values, color="skyblue")

ax.set_title("Average Fare by Service Type", fontsize=13, pad=12)
ax.set_xlabel("Service Type", fontsize=11)
ax.set_ylabel("Average Fare", fontsize=11)
ax.grid(axis="y", linestyle="--", alpha=0.4)

# Add value labels on top of each bar
ax.bar_label(bars, fmt="%.2f", padding=3, fontsize=9)

# plt.tight_layout()
plt.show()
```
